```python
import os
import tensorflow as tf
from tqdm import tqdm
```

```python
# Paths
raw_data_path = r"C:\Users\Legion\Desktop\pneumo\data\chest_xray\chest_xray"
processed_data_path = r"C:\Users\Legion\Desktop\pneumo\data\processed"
```

```python
# Create directories for processed data
for split in ['train', 'test', 'val']:
    for class_name in ['NORMAL', 'PNEUMONIA']:
        path = os.path.join(processed_data_path, split, class_name)
        os.makedirs(path, exist_ok=True)
```

```python
# Function to preprocess and save images
def preprocess_and_save_images(source_dir, target_dir, img_size=(224, 224)):
    for class_name in tqdm(os.listdir(source_dir), desc=f"Processing {os.path.basename(source_dir)}"):
        class_source_path = os.path.join(source_dir, class_name)
        class_target_path = os.path.join(target_dir, class_name)
        os.makedirs(class_target_path, exist_ok=True)
        
        for file_name in os.listdir(class_source_path):
            try:
                # Load, resize, and normalize image
                img_path = os.path.join(class_source_path, file_name)
                img = tf.keras.utils.load_img(img_path, target_size=img_size)
                img_array = tf.keras.utils.img_to_array(img) / 255.0  # Normalize to [0, 1]
                
                # Save preprocessed image
                save_path = os.path.join(class_target_path, file_name)
                tf.keras.utils.save_img(save_path, img_array)
            except Exception as e:
                print(f"Error processing {img_path}: {e}")

# Apply preprocessing to all splits
for split in ['train', 'test', 'val']:
    preprocess_and_save_images(
        source_dir=os.path.join(raw_data_path, split),
        target_dir=os.path.join(processed_data_path, split)
    )

print("Preprocessing complete. Data saved to:", processed_data_path)
```
